```python
import pandas as pd
import numpy as np

# Create a sample dataset
data = {
    'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Ezhil', 'Fathima'],
    'Age': [20, 21, np.nan, 22, 19, 21],
    'Department': ['CSE', 'ECE', 'CSE', 'EEE', 'CSE', 'ECE'],
    'Marks': [85, 72, 90, np.nan, 65, 78]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Column filter
filtered_columns = df[['Name', 'Age', 'Department', 'Marks']]

# Row filter
row_filtered = filtered_columns[
    filtered_columns['Marks'].fillna(0) >= 70
].copy()

# Handle missing values
row_filtered['Age'] = row_filtered['Age'].fillna(
    row_filtered['Age'].median()
)

row_filtered['Marks'] = row_filtered['Marks'].fillna(
    row_filtered['Marks'].mean()
)

# Remove duplicate rows
cleaned_data = row_filtered.drop_duplicates()

print("\nCleaned Dataset:")
print(cleaned_data)
```
