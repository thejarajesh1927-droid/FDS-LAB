```python
import pandas as pd
```

```python
# Create student dataset
data = {
'Student_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Ezhil',
'Fathima', 'Gokul', 'Harini', 'Ishaan', 'Jaya'],
'Department': ['CSE', 'IT', 'CSE', 'ECE', 'IT',
'CSE', 'ECE', 'IT', 'CSE', 'ECE'],
'Gender': ['M', 'M', 'F', 'F', 'M',
'F', 'M', 'F', 'M', 'F'],
'Attendance': [92, 78, 88, 95, 82, 90, 76, 94, 85, 89],
'Internal': [42, 35, 39, 45, 36, 41, 32, 44, 38, 40],
'External': [44, 38, 41, 46, 40, 43, 35, 45, 39, 42],
'Study_Hours': [6, 4, 5, 7, 5, 6, 3, 7, 4, 5]
}
df = pd.DataFrame(data)
print("ORIGINAL DATASET")
print(df)
```

```python
# 1. DATAFRAME INFORMATION
print("\nDATAFRAME INFORMATION")
print("Shape:", df.shape)
print("Columns:", list(df.columns))
print("\nData Types:")
print(df.dtypes)
```

```python
# 2. COLUMN SELECTION
print("\nSELECTED COLUMNS")
print(df[['Name', 'Department', 'Attendance', 'Study_Hours']])
```

```python
# 3. INDEXING USING LOC
print("\nLOC - STUDENTS FROM ROW 2 TO 5")
print(df.loc[2:5, ['Name', 'Department', 'Internal', 'External']])
```

```python
# 4. INDEXING USING ILOC
print("\nILOC - FIRST FIVE RECORDS AND SELECTED COLUMNS")
print(df.iloc[:5, [1, 2, 4, 5, 6]])
```

```python

# 5. FILTERING
# Students with attendance >= 90
high_attendance = df[df['Attendance'] >= 90]

print("\nSTUDENTS WITH ATTENDANCE >= 90")
print(high_attendance[['Student_ID', 'Name', 'Department', 'Attendance']])


# Students with external marks >= 42
high_external = df[df['External'] >= 42]

print("\nSTUDENTS WITH EXTERNAL MARKS >= 42")
print(high_external[['Student_ID', 'Name', 'External']])


# Students satisfying multiple performance conditions
high_performers = df[
    (df['Attendance'] >= 85) &
    (df['Internal'] >= 38) &
    (df['External'] >= 40)
]

print("\nSTUDENTS SATISFYING ALL PERFORMANCE CONDITIONS")
print(high_performers[
    ['Student_ID', 'Name', 'Department', 'Attendance', 'Internal', 'External']
])
```

```python
# 6. CREATE NEW COLUMNS
df['Total'] = df['Internal'] + df['External']
df['Percentage'] = (df['Total'] / 100) * 100
print("\nDATASET AFTER CALCULATING TOTAL AND PERCENTAGE")
print(df[['Student_ID', 'Name', 'Internal', 'External', 'Total', 'Percentage']])
```

```python
# 7. PERFORMANCE CATEGORY
def performance_category(percentage):
    if percentage >= 85:
       return 'Excellent'
    elif percentage >= 75:
       return 'Very Good'
    elif percentage >= 60:
       return 'Good'
    else:
       return 'Needs Improvement'
df['Performance'] = df['Percentage'].apply(performance_category)
print("\nPERFORMANCE CATEGORY")
print(df[['Student_ID', 'Name', 'Percentage', 'Performance']])
```

```python
# 8. SORTING
sorted_df = df.sort_values(
by='Percentage',
ascending=False
)
print("\nSTUDENTS SORTED BY PERCENTAGE")
print(sorted_df[['Student_ID', 'Name', 'Department', 'Percentage', 'Performance']])
# 9. TOP PERFORMERS
print("\nTOP 3 PERFORMERS")
print( df.nlargest(
3,
'Percentage'
)[['Student_ID', 'Name', 'Department', 'Percentage']]
)
# 10. DEPARTMENT-WISE GROUPING
department_summary = df.groupby('Department').agg(
Students=('Student_ID', 'count'),
Average_Attendance=('Attendance', 'mean'), Average_Internal=('Internal', 'mean'),
Average_External=('External', 'mean'),
Average_Percentage=('Percentage', 'mean'),
Highest_Percentage=('Percentage', 'max'),
Lowest_Percentage=('Percentage', 'min')
)
print("\nDEPARTMENT-WISE SUMMARY")
print(department_summary.round(2))
# 11. DEPARTMENT AND GENDER GROUPING
gender_department = df.groupby( ['Department', 'Gender']
).agg(
Student_Count=('Student_ID', 'count'),
Average_Percentage=('Percentage', 'mean'),
Average_Attendance=('Attendance', 'mean')
)
print("\nDEPARTMENT AND GENDER-WISE SUMMARY")
print(gender_department.round(2))
# 12. PERFORMANCE CATEGORY COUNT
performance_count = df.groupby( 'Performance'
).size().sort_values(ascending=False)
print("\nPERFORMANCE CATEGORY COUNT")
print(performance_count)
# 13. DEPARTMENT-WISE STUDY HOURS
study_summary = df.groupby('Department').agg(
Average_Study_Hours=('Study_Hours', 'mean'),
Maximum_Study_Hours=('Study_Hours', 'max'),
Minimum_Study_Hours=('Study_Hours', 'min')
)
print("\nDEPARTMENT-WISE STUDY HOURS")
print(study_summary.round(2))
# 14. CREATE DEPARTMENT DETAILS
department_data = {
'Department': ['CSE', 'IT', 'ECE'],
'Faculty_Count': [12, 10, 8],
'Department_HOD': [
'Dr. Kumar',
'Dr. Priya',
'Dr. Ravi'
]
}
department_df = pd.DataFrame(department_data)
print("\nDEPARTMENT DETAILS")
print(department_df)
# 15. MERGE DATAFRAMES
merged_df = pd.merge( df, department_df, on='Department', how='left'
)
print("\nMERGED DATASET")
print(
merged_df[
['Student_ID', 'Name', 'Department',
'Percentage', 'Faculty_Count', 'Department_HOD']
]
)
# ---------------------------------------------------------
# 16. PIVOT TABLE
# ---------------------------------------------------------
pivot_table = pd.pivot_table(
df, values='Percentage', index='Department', columns='Gender',
aggfunc='mean'
)
print("\nPIVOT TABLE - AVERAGE PERCENTAGE")
print(pivot_table.round(2))
# ---------------------------------------------------------
# 17. OVERALL STATISTICS
# ---------------------------------------------------------
print("\nOVERALL STATISTICS")
print("Average Attendance:", round(df['Attendance'].mean(), 2))
print("Average Internal Marks:", round(df['Internal'].mean(), 2))
print("Average External Marks:",
round(df['External'].mean(), 2))
print("Average Percentage:", round(df['Percentage'].mean(), 2))
print("Highest Percentage:", df['Percentage'].max())
print("Lowest Percentage:", df['Percentage'].min())
# ---------------------------------------------------------
# 18. FINAL DATASET
# ---------------------------------------------------------
print("\nFINAL MANIPULATED DATASET")
print(
df[
['Student_ID', 'Name', 'Department', 'Gender',
'Attendance', 'Internal', 'External',
'Study_Hours', 'Total', 'Percentage', 'Performance'] ]
)
```
