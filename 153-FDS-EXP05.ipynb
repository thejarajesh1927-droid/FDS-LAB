```python
# ---------------------------------------------------------
# 1. NORMAL DISTRIBUTION
# ---------------------------------------------------------

import numpy as np
from scipy import stats

print("=" * 65)
print("\tANALYSIS OF PROBABILITY DISTRIBUTIONS")
print("=" * 65)

# Parameters of Normal Distribution
normal_mean = 50
normal_std = 10

# Create a Normal distribution object
normal_dist = stats.norm(loc=normal_mean, scale=normal_std)

# Generate values for plotting
normal_x = np.linspace(
    normal_mean - 4 * normal_std,
    normal_mean + 4 * normal_std,
    200
)

# Calculate PDF values
normal_pdf = normal_dist.pdf(normal_x)

# Calculate CDF values
normal_cdf = normal_dist.cdf(normal_x)

# Probability that X is less than or equal to 60
p_normal_less_60 = normal_dist.cdf(60)

# Probability that X is greater than 40
p_normal_greater_40 = 1 - normal_dist.cdf(40)

# Probability that X lies between 40 and 60
p_normal_between = normal_dist.cdf(60) - normal_dist.cdf(40)

# Statistical measures
normal_variance = normal_dist.var()
normal_std_deviation = normal_dist.std()

# Display results
print("\n1. NORMAL DISTRIBUTION")
print("-" * 65)

print(f"Mean                : {normal_mean:.2f}")
print(f"Standard Deviation  : {normal_std:.2f}")
print(f"Variance             : {normal_variance:.2f}")
print(f"P(X <= 60)          : {p_normal_less_60:.4f}")
print(f"P(X > 40)           : {p_normal_greater_40:.4f}")
print(f"P(40 < X < 60)      : {p_normal_between:.4f}")
```

```python
# ---------------------------------------------------------
# 2. BINOMIAL DISTRIBUTION
import numpy as np
from scipy import stats

# Parameters of Binomial Distribution
binomial_n = 10
binomial_p = 0.5

# Create a Binomial distribution object
binomial_dist = stats.binom(
    n=binomial_n,
    p=binomial_p
)

# Possible number of successes
binomial_x = np.arange(0, binomial_n + 1)

# Calculate PMF values
binomial_pmf = binomial_dist.pmf(binomial_x)

# Calculate CDF values
binomial_cdf = binomial_dist.cdf(binomial_x)

# Probability of exactly 6 successes
p_binomial_exact_6 = binomial_dist.pmf(6)

# Probability of at most 6 successes
p_binomial_at_most_6 = binomial_dist.cdf(6)

# Probability of more than 6 successes
p_binomial_more_6 = 1 - binomial_dist.cdf(6)

# Statistical measures
binomial_mean = binomial_dist.mean()
binomial_variance = binomial_dist.var()
binomial_std_deviation = binomial_dist.std()

# Display results
print("\n2. BINOMIAL DISTRIBUTION")
print("-" * 65)

print(f"Number of Trials (n)      : {binomial_n}")
print(f"Probability of Success    : {binomial_p:.2f}")
print(f"Mean                      : {binomial_mean:.2f}")
print(f"Variance                  : {binomial_variance:.2f}")
print(f"Standard Deviation        : {binomial_std_deviation:.2f}")
print(f"P(X = 6)                  : {p_binomial_exact_6:.4f}")
print(f"P(X <= 6)                 : {p_binomial_at_most_6:.4f}")
print(f"P(X > 6)                  : {p_binomial_more_6:.4f}")
```

```python
# ---------------------------------------------------------
# 3. POISSON DISTRIBUTION
# ---------------------------------------------------------

import numpy as np
from scipy import stats

# Parameter of Poisson Distribution
poisson_lambda = 4

# Create a Poisson distribution object
poisson_dist = stats.poisson(mu=poisson_lambda)

# Possible number of occurrences
poisson_x = np.arange(0, 13)

# Calculate PMF values
poisson_pmf = poisson_dist.pmf(poisson_x)

# Calculate CDF values
poisson_cdf = poisson_dist.cdf(poisson_x)

# Probability of exactly 5 occurrences
p_poisson_exact_5 = poisson_dist.pmf(5)

# Probability of at most 5 occurrences
p_poisson_at_most_5 = poisson_dist.cdf(5)

# Probability of more than 5 occurrences
p_poisson_more_5 = 1 - poisson_dist.cdf(5)

# Statistical measures
poisson_mean = poisson_dist.mean()
poisson_variance = poisson_dist.var()
poisson_std_deviation = poisson_dist.std()

# Display results
print("\n3. POISSON DISTRIBUTION")
print("-" * 65)

print(f"Average Rate (lambda)     : {poisson_lambda:.2f}")
print(f"Mean                      : {poisson_mean:.2f}")
print(f"Variance                  : {poisson_variance:.2f}")
print(f"Standard Deviation        : {poisson_std_deviation:.2f}")
print(f"P(X = 5)                  : {p_poisson_exact_5:.4f}")
print(f"P(X <= 5)                 : {p_poisson_at_most_5:.4f}")
print(f"P(X > 5)                  : {p_poisson_more_5:.4f}")
```

```python
# ---------------------------------------------------------
# 4. DISPLAY PROBABILITY DISTRIBUTION GRAPHS
# ---------------------------------------------------------

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 9))

# Normal Distribution
plt.subplot(3, 1, 1)

plt.plot(
    normal_x,
    normal_pdf,
    linewidth=2,
    label="Normal PDF"
)

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.grid(True, alpha=0.3)
plt.legend()


# Binomial Distribution
plt.subplot(3, 1, 2)

plt.stem(
    binomial_x,
    binomial_pmf,
    basefmt=" "
)

plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.grid(True, alpha=0.3)


# Poisson Distribution
plt.subplot(3, 1, 3)

plt.stem(
    poisson_x,
    poisson_pmf,
    basefmt=" "
)

plt.title("Poisson Distribution")
plt.xlabel("Number of Occurrences")
plt.ylabel("Probability")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 5. DISPLAY SELECTED PMF VALUES
# ---------------------------------------------------------

print("\nSELECTED PROBABILITY VALUES")
print("=" * 65)

# Binomial Distribution PMF
print("\nBinomial Distribution PMF:")
print("Number of Successes    Probability")

for x, probability in zip(binomial_x, binomial_pmf):
    print(f"{x:20d} {probability:.4f}")


# Poisson Distribution PMF
print("\nPoisson Distribution PMF:")
print("Number of Occurrences  Probability")

for x, probability in zip(poisson_x, poisson_pmf):
    print(f"{x:20d} {probability:.4f}")


print("\n" + "=" * 65)
print("Analysis of probability distributions completed successfully.")
print("=" * 65)
```
